```python
#1.List 3 differences between Reinforcement Learning, Supervised Learning, and Unsupervised Learning in a table, using real
# examples from apps like Instagram (feed ranking), Spotify (music recommendations), and chess game AI.

Reinforcement Learning
Learns by taking actions and receiving rewards or penalties.
Feedback comes as a reward or penalty after an action.
Chess AI: chooses a move and receives a positive reward for winning or a negative reward for losing.

Supervised Learning
Learns from labeled data where the correct answer is already provided.
Feedback is the actual/correct label.
Instagram feed ranking: a model can learn from labeled examples of posts users liked or disliked.

Unsupervised Learning
Learns patterns and structures from unlabeled data.
There is no predefined correct answer.
Spotify: can group songs or users with similar listening patterns without predefined categories.
```

```python
#2.Pick any one real-world application of Reinforcement Learning (such as game AI, robotics, self-driving cars,
# or recommendation systems) and explain step-by-step how an agent, environment, actions, and rewards would work in that 
# scenario.<br><br><em><strong>Hint:</strong> Use your own words and pick an example you find interesting, like how a Zomato
# recommendation engine could use RL to suggest restaurants.</em>

A good real-world example of Reinforcement Learning is a restaurant recommendation system used in a food delivery app such as Zomato.
The system can learn from a user's actions and gradually improve the restaurants it recommends.

Step 1: Agent
The agent is the AI recommendation system. Its job is to decide which restaurant or food option should be recommended to the user.
    
Step 2: Environment
The environment is the food delivery app and the user's interaction with it. It includes information such as:
- User's previous orders
- Restaurant ratings
- User's location
- Food preferences
- Time of day
- Restaurant availability

Step 3: State
The agent first observes the current state of the user.
For example:
The user is looking for dinner, usually orders pizza, is located near Vadodara, and has previously liked highly-rated restaurants.

This information represents the current state.
    
Step 4: Action
Based on the state, the agent chooses an action. For example, it can recommend:
- Pizza restaurant
- South Indian restaurant
- Chinese restaurant
- Fast-food restaurant
So, the action is the restaurant recommendation made by the agent.
```
